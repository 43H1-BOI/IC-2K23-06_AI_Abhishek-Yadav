# Operations on NumPy Array

## Reshaping of NumPy Array

```python
import numpy as np
```

```python
# 2D array with shape ( 2 , 3 )
arr2d = np.array([[1,2,3],[4,5,6]])
print(arr2d,arr2d.shape)
```

### Converting it to a 1D Array

```python
arr1d = arr2d.flatten()
print(arr1d,arr1d.shape)
```

### Converting it to a ( 3 , 2 ) Array

```python
reshaped = arr2d.reshape((3,2))
print(reshaped,reshaped.shape)
```

```python
reshaped = arr1d.reshape((2,3))
print(reshaped,reshaped.shape)
```
