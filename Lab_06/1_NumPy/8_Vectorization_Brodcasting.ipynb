# Vectorization vs Brodcasting in NumPy

```python
import numpy as np
```

### Vectorization refers to the process in which the *ndarray* can process an operation without iterating over whole array

```python
arr = np.array([1,2,3,4,5])

# this is done using vectorization
print(arr ** 2)
```

### Brodcasting is conversion of non compatible object i.e. scalar to vector
- In Above Example : arr is **vector** and 2 is **scalar** means a single value

```python
# arr + 5
# [1,2,3,4,5] + 5
# [1,2,3,4,5] + [5,5,5,5,5]

# The Above is working of Brodcasting but does not uses extra memory
```
