```python
import seaborn as sns
import matplotlib.pyplot as plt
```

## Importing Sample Data

```python
penguins = sns.load_dataset("penguins")
penguins.sample(5)
```

## Distribution Plot :- Graphs which defines distributiion of data
### Ex. Histogram
- ### x -> Value
- ### bins -> no of partitions ( we want for data to be divided into )

```python
sns.histplot(
    data = penguins,
    x = "body_mass_g",
    bins = 50
)
```
