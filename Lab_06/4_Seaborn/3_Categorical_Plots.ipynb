```python
import seaborn as sns
import matplotlib.pyplot as plt
```

## Importing Sample Data

```python
tips = sns.load_dataset("tips")
flights = sns.load_dataset("flights")
```

## Categorical Plot :- Graphs which defines categories
### Ex. Box , Bar Plot
- ### x -> Category
- ### y -> Value


### Box Plot

```python
tips.sample(5)
```

```python
sns.relplot(
    data = tips,
    x = "total_bill",
    y = "tip",
    hue = "sex"
)
```

#### Using `boxplot()`

```python
sns.boxplot(
    data = tips,
    x = "day",
    y = "tip",
    hue = "sex"
)
```

### Bar Plot

#### Using `barplot()`

```python
flights.sample(5)
```

```python
# the shaded part is known as confidence interval ( default value = 95% ) 
sns.barplot(
    data = flights,
    x = "year",
    y = "passengers",
    hue = "month"
)
```
