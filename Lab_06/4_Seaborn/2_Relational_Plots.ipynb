```python
import seaborn as sns
import matplotlib.pyplot as plt
```

## Importing Sample Data

```python
tips = sns.load_dataset("tips")
flights = sns.load_dataset("flights")
```

## Relational Plots :- Graph which shows relation b/w two variables
### Ex. Scatter *(Default)* , Line Plot
- ### x -> Independent Variable
- ### y -> Dependent Variable


### Scatter Plot

```python
tips.sample(5)
```

#### Using `relplot()`

```python
sns.relplot(
    data = tips,
    x = "total_bill",
    y = "tip",
    hue = "sex"
)
```

#### Using `scatterplot()`

```python
sns.scatterplot(
    data = tips,
    x = "total_bill",
    y = "tip",
    hue = "sex"
)
```

### Line Plot

#### Using `relplot(kind = "line")`

```python
x_vals = range(1,101)
y_vals = [(-i**(-1/2)) for i in x_vals]
```

```python
sns.relplot(
    x = x_vals,
    y = y_vals,
    kind = "line"
)
```

#### Using `lineplot()`

```python
flights.sample(5)
```

```python
# the shaded part is known as confidence interval ( default value = 95% ) 
sns.lineplot(
    data = flights,
    x = "month",
    y = "passengers"
)
```
