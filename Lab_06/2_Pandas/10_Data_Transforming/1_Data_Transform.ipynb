# Data Transforming ( Feature Engineering ) Pt.1

```python
import pandas as pd
```

```python
# importing data
data = pd.read_csv("../Res/9_data.csv")
data.head(3)
```

## .apply(fxn) -> takes function and apply it to whole column ( data transformation )

```python
data["tax"] = data["income"].apply(lambda income : "10%" if income >= 50_000 else "20%" )

data
```

## .map() -> to convert one value to another

```python
gender_map = {"Male":"M" , "Female":"F" , "Unknown":"U"}

data["gender"] = data["gender"].map(gender_map)

data
```

## .assign() -> to update values

```python
# incrementing value by 10%
# and assigning it to new_income
data = data.assign(new_income = data["income"] * 1.1)

data
```

## .replace(old , new) -> to replace value from old to new

```python
data["country"] = data["country"].replace("USA","United States")

data
```
