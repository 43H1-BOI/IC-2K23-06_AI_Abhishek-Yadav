# Groupby and Aggrigate Functions

```python
import pandas as pd
```

```python
# importing data
data = pd.read_csv("../Res/cleaned_data.csv")
data
```

## .groupby() -> Used to group rows having same value
## Aggregate Functions :- agg() or aggregate()
- ### sum() -> returns sum of values
- ### count() -> returns count of values
- ### max() -> returns max value
- ### min() -> returns min value


```python
# group by "country" and give mean for "income"
data.groupby("country")["income"].mean()
```

```python
# group by "country" and give mean , median , sum , min , max for "income"
data.groupby("country")["income"].agg(["mean","median","min","max","sum"])
```

```python
# change these column names
data.groupby("country")["income"].agg(Mean = "mean",Median = "median").rename_axis("Country")
```

```python
# To apply different functions on different columns
data.groupby("country").agg({
    "income" : "mean",
    "age" : "median"
})
```

```python
# To Rename these column name
data.groupby("country").agg(
    # new_col_name = ("col","agg_fun")
    mean_income = ("income","mean"),
    median_age = ("age","median")
)
```
