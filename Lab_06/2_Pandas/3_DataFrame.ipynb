```python
import numpy as np
import pandas as pd
```

```python
# Without Labels
data = {
    "Name" : ["Abhi","Anu","XYZ"],
    "Age" : [22,18,12]
}

df = pd.DataFrame(data)
df
```

```python
# With Labels
values = ["1st Rank","2nd Rank","3rd Rank"]
df = pd.DataFrame(data,index = values)
df
```

```python
# Using Numpy Array


nparray = np.array([["Abhi","Anu","XYZ"],[22,18,12]])

# columns is compulsory for nparray -> DataFrame
df = pd.DataFrame(nparray,index = ["Name","Age"],columns = values)
df
```
