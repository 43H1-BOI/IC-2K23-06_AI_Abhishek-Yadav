### We Can Handle NaN Values Using Following Methods :
- #### Drop Rows / Columns
    - Using fillna(value)
    - Using ffill() -> Used to fill prev. value to next NaN Value
    - Using bfill() -> Used to fill next value to preceding NaN Value

### fillna(value) method

```python
# Filling Data with Specific Values
data.fillna(0)
```

```python
# We can fill up data using this method for best results
mean = data["age"].mean()
median = data["age"].median()

# this will change all null to mean
data.fillna(mean)
```

```python
# Copying data to another DataFrame
new_data = data.copy()

# changing only age column
new_data["age"] = data["age"].fillna(median)
```

```python
new_data
```

### ffill() method - This will fill up null value with its prev value

```python
f_data = data.copy()
```

```python
# This will fill up null value with its prev value
f_data["age"] = f_data["age"].ffill()
```

```python
f_data
```

### bfill() method - This will fill up null value with its next value

```python
b_data = data.copy()
```

```python
# This will fill up null value with its next value
b_data["age"] = b_data["gender"].bfill()
```

```python
# But if last value is null then it will remain null as there is no value after that

b_data
```

### dropna() method - This will drop all values having null

```python
n_data = data.copy()
```

```python
# This will drop all values having null
n_data = n_data.dropna()
```
