# Handling Duplicates

```python
import pandas as pd
```

```python
data = pd.read_csv("../Res/9_data.csv")
```

```python
data
```

- ## .duplicated() method is used to find duplicates

```python
# checking for whole data
data.duplicated()

data.duplicated().head()
```

```python
# Checking for specific Columns
data["age"].duplicated()

data["age"].duplicated().head()
```

```python
# Checking for multiple columns
data[["age","country"]].duplicated()

data[["age","country"]].duplicated().head()
```

- ## .drop_duplicates() - To drop all duplicate rows
- ## .drop_duplicates(inplace = True) - modifies original data 

```python
d2 = data.drop_duplicates()
d2
```

```python
data
```

```python
data.drop_duplicates(inplace = True)
```
