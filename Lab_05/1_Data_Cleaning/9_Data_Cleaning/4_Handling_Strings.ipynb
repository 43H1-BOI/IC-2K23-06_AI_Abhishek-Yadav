### split() method

```python
data["name"].str.split(" ")
```

```python
# Let's Do Something Else
# We'll make two columns named f_name and l_name

data
```

```python
data[["f_name","l_name"]] = data["name"].str.split(" ", n = 1 , expand = True)
# step by step breakdown
# data[["f_name","l_name"]] -> to store data returning into 2 diff columns
# .split()
    # " " -> to split from " "
    # n = 1 -> will split only once so that no middle name will harm the code
    # expand = True -> so that it will returns value in form of columns not list

# demonstration of expand = True
data["name"].str.split(" ", n = 1 , expand = True)
```
