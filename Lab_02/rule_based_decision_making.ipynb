## 4. Rule-Based Decision Making (Production System)
### Formulation
- Simple forward-chaining diagnostic engine evaluating user inputs against registered rule conditions.

```python
def rule_based_diagnostic(facts):
    """
    Simple rule engine evaluating symptoms to recommend an action.
    """
    rules = [
        {
            "conditions": {"fever": True, "cough": True, "shortness_of_breath": True},
            "decision": "High risk respiratory infection -> Consult hospital emergency immediately."
        },
        {
            "conditions": {"fever": True, "cough": True, "shortness_of_breath": False},
            "decision": "Common viral flu -> Rest, hydrate, and take antipyretics."
        },
        {
            "conditions": {"fever": False, "cough": True, "shortness_of_breath": False},
            "decision": "Mild throat irritation/allergy -> Warm fluids and lozenges."
        },
        {
            "conditions": {"fever": False, "cough": False, "fatigue": True},
            "decision": "Exhaustion or mild stress -> Ensure 8 hours of sleep and balanced diet."
        }
    ]
    
    print(f"Input Patient Facts: {facts}\n")
    
    for i, rule in enumerate(rules, start=1):
        # Check if all conditions in this rule match the provided facts
        match = True
        for key, expected_val in rule["conditions"].items():
            if facts.get(key, None) != expected_val:
                match = False
                break
        
        if match:
            return f"Triggered Rule #{i}: {rule['decision']}"
            
    return "No exact rule match. Clinical observation recommended."
```

```python
patient_symptoms = {"fever": True, "cough": True, "shortness_of_breath": False}
result = rule_based_diagnostic(patient_symptoms)
print("Recommendation:", result)
```
