## 3. Missionaries and Cannibals Problem
### Formulation
- **State:** `(m, c, b)` where `m` = missionaries on left, `c` = cannibals on left, `b` = boat position (`1` for left, `0` for right).
- **Rule:** Missionaries are eaten if $0 < m < c$ on either bank.
- **Allowed Boat Moves:** $(1,0), (2,0), (0,1), (0,2), (1,1)$.

```python
from collections import deque

def is_valid(m, c):
    # Check bounds (0 to 3)
    if not (0 <= m <= 3 and 0 <= c <= 3):
        return False
    # Left bank safety check
    if m > 0 and m < c:
        return False
    # Right bank safety check
    m_right, c_right = 3 - m, 3 - c
    if m_right > 0 and m_right < c_right:
        return False
    return True
```

```python
def missionaries_and_cannibals():
    start_state = (3, 3, 1)  # Left bank: 3M, 3C, boat=1
    goal_state = (0, 0, 0)   # Left bank: 0M, 0C, boat=0
    
    # Queue holds: (current_state, path)
    queue = deque([(start_state, [start_state])])
    visited = {start_state}
    
    # Possible boat passengers: (M, C)
    moves = [(1, 0), (2, 0), (0, 1), (0, 2), (1, 1)]
    
    while queue:
        (m, c, b), path = queue.popleft()
        
        if (m, c, b) == goal_state:
            return path
        
        for dm, dc in moves:
            if b == 1:  # Boat crosses Left -> Right
                new_state = (m - dm, c - dc, 0)
            else:       # Boat crosses Right -> Left
                new_state = (m + dm, c + dc, 1)
                
            if is_valid(new_state[0], new_state[1]) and new_state not in visited:
                visited.add(new_state)
                queue.append((new_state, path + [new_state]))
                
    return None
```

```python
path = missionaries_and_cannibals()
if path:
    print(f"Found solution in {len(path) - 1} boat crossings:\n")
    for i, (m, c, b) in enumerate(path):
        boat_loc = "Left" if b == 1 else "Right"
        print(f"Step {i:2d}: Left Bank = ({m}M, {c}C) | Right Bank = ({3-m}M, {3-c}C) | Boat = {boat_loc}")
```
