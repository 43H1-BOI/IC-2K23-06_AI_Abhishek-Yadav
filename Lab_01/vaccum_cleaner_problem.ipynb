## 1. Vacuum Cleaner Problem
### Formulation
- **State**: `(Agent_Location, Status_A, Status_B)`
- **Actions**: `SUCK`, `MOVE_LEFT`, `MOVE_RIGHT`
- **Rules**:
  - IF room is `Dirty` $\rightarrow$ `SUCK`
  - IF room is `Clean` and at `A` $\rightarrow$ Move `RIGHT`
  - IF room is `Clean` and at `B` $\rightarrow$ Move `LEFT`

```python
def vacuum_cleaner(start_room='A', status_a='Dirty', status_b='Dirty'):
    rooms = {'A': status_a, 'B': status_b}
    current = start_room
    step = 1
    
    print(f"Initial State: Agent at {current} | Rooms: {rooms}")
    
    while rooms['A'] == 'Dirty' or rooms['B'] == 'Dirty':
        if rooms[current] == 'Dirty':
            rooms[current] = 'Clean'
            print(f"Step {step}: Room {current} is Dirty -> Action: SUCK")
        else:
            next_room = 'B' if current == 'A' else 'A'
            action = 'MOVE RIGHT' if current == 'A' else 'MOVE LEFT'
            print(f"Step {step}: Room {current} is Clean -> Action: {action}")
            current = next_room
        step += 1
        
    print(f"Done! Both rooms clean in {step - 1} steps. Final: {rooms}")
```

```python

# Run
vacuum_cleaner(start_room='A', status_a='Dirty', status_b='Dirty')
```
