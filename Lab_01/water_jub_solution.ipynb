# Water Jug Problem (Easy Simulation Method)

### Rule-based Approach:
Instead of searching through a tree or graph, we follow a simple repeating strategy:
1. **Fill Jug 1** whenever it is empty.
2. **Pour from Jug 1 to Jug 2**.
3. **Empty Jug 2** whenever it is full.
4. Stop when either jug reaches the target amount.

```python
import math

def water_jug(cap1, cap2, target):
    # Quick check: is it mathematically possible?
    if target > max(cap1, cap2) or target % math.gcd(cap1, cap2) != 0:
        print("No solution possible.")
        return

    j1 = 0  # current water in Jug 1
    j2 = 0  # current water in Jug 2
    step = 1

    print(f"Initial State: Jug1 = {j1}L, Jug2 = {j2}L\n")

    while j1 != target and j2 != target:
        # Rule 1: If Jug 1 is empty, fill it
        if j1 == 0:
            j1 = cap1
            print(f"Step {step}: Fill Jug 1 -> ({j1}L, {j2}L)")
        
        # Rule 2: If Jug 2 is full, empty it
        elif j2 == cap2:
            j2 = 0
            print(f"Step {step}: Empty Jug 2 -> ({j1}L, {j2}L)")
        
        # Rule 3: Otherwise, pour from Jug 1 into Jug 2
        else:
            pour_amount = min(j1, cap2 - j2)
            j1 -= pour_amount
            j2 += pour_amount
            print(f"Step {step}: Pour Jug 1 to Jug 2 -> ({j1}L, {j2}L)")
        
        step += 1

    print(f"\nTarget reached in {step - 1} steps!")
```

### Run the Code
Example: Jug 1 = **4L**, Jug 2 = **3L**, Target = **2L**.

```python
water_jug(cap1=4, cap2=3, target=2)
```
